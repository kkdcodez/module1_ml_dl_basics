# 09-1 추가 실습 — RNN 은닉 상태 계산

### 1. PyTorch 불러오기

```python
import torch
```

### 2. 가중치와 입력 시퀀스 준비하기

```python
w_x = torch.tensor(1.0)
w_h = torch.tensor(0.5)
b = torch.tensor(0.0)

h = torch.tensor(0.0)
x_seq = torch.tensor([1.0, 0.0])
```

### 3. 타임스텝별 은닉 상태 계산하기

```python
for t, x_t in enumerate(x_seq, start=1):
    h = torch.tanh(x_t * w_x + h * w_h + b)
    print(f"t={t}, 은닉 상태={h.item():.4f}")
```

### 4. 입력 순서를 바꾸어 결과 비교하기

```python
def run_sequence(values):
    h = torch.tensor(0.0)

    for x in values:
        x_t = torch.tensor(float(x))
        h = torch.tanh(x_t * w_x + h * w_h + b)

    return h.item()

print("[1, 0]:", round(run_sequence([1, 0]), 4))
print("[0, 1]:", round(run_sequence([0, 1]), 4))
```
